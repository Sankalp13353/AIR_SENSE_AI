```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")
```

```python
df = pd.read_csv("aqi_india_38cols_knn_final.csv")

df.head()
```

```python
cleaned_df = df.copy()
```

```python
pollutants = [
    "pm2_5_ugm3",
    "pm10_ugm3",
    "co_ugm3",
    "no2_ugm3",
    "so2_ugm3",
    "o3_ugm3",
    "dust_ugm3"
]

cleaned_df[pollutants].describe()
```

```python
bool_cols = cleaned_df.select_dtypes(include="bool").columns

for col in bool_cols:
    print(col)
    print(cleaned_df[col].value_counts())
    print("-"*40)
```

```python
cleaned_df["datetime"] = pd.to_datetime(cleaned_df["datetime"])
```

```python
cleaned_df.dtypes
```

```python
cleaned_df = cleaned_df.sort_values("datetime")
```

```python
cleaned_df.reset_index(drop=True, inplace=True)
```

```python
print(cleaned_df.shape)

print(cleaned_df.isnull().sum().sum())

print(cleaned_df.duplicated().sum())
```

```python
cleaned_df.groupby("city")["state"].nunique().sort_values(ascending=False)
```

```python
cleaned_df[
    (cleaned_df["latitude"] < 6) |
    (cleaned_df["latitude"] > 38) |
    (cleaned_df["longitude"] < 68) |
    (cleaned_df["longitude"] > 98)
]
```

```python
for col in pollutants:
    print(col, (cleaned_df[col] < 0).sum())
```

```python
(cleaned_df["us_aqi"] < 0).sum()
```

```python
cleaned_df.to_csv(
    "cleaned_air_quality.csv",
    index=False
)
```
