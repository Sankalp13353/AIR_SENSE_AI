```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("cleaned_air_quality.csv")

df.head()
```

```python
df["datetime"] = pd.to_datetime(df["datetime"])
```

```python
df = df.sort_values(["city", "datetime"])

df.reset_index(drop=True, inplace=True)
```

```python
df["hour"] = df["datetime"].dt.hour

df["day"] = df["datetime"].dt.day

df["year"] = df["datetime"].dt.year

df[["datetime", "hour", "day", "year"]].head()
```

```python
df["aqi_lag_1"] = (
    df.groupby("city")["us_aqi"]
      .shift(1)
)
```

```python
df["aqi_lag_3"] = (
    df.groupby("city")["us_aqi"]
      .shift(3)
)
```

```python
df["aqi_lag_6"] = (
    df.groupby("city")["us_aqi"]
      .shift(6)
)
```

```python
df["aqi_lag_12"] = (
    df.groupby("city")["us_aqi"]
      .shift(12)
)
```

```python
df["aqi_lag_24"] = (
    df.groupby("city")["us_aqi"]
      .shift(24)
)
```

```python
df[
    [
        "city",
        "datetime",
        "us_aqi",
        "aqi_lag_1",
        "aqi_lag_24"
    ]
].head(30)
```

```python
df[
    [
        "city",
        "datetime",
        "us_aqi",
        "aqi_lag_1",
        "aqi_lag_3",
        "aqi_lag_6",
        "aqi_lag_12",
        "aqi_lag_24"
    ]
].head(30)
```

```python
df["aqi_roll_mean_24"] = (
    df.groupby("city")["us_aqi"]
      .transform(lambda x: x.rolling(window=24).mean())
)
```

```python
df["aqi_roll_max_24"] = (
    df.groupby("city")["us_aqi"]
      .transform(lambda x: x.rolling(window=24).max())
)
```

```python
df["aqi_roll_min_24"] = (
    df.groupby("city")["us_aqi"]
      .transform(lambda x: x.rolling(window=24).min())
)
```

```python
df["aqi_roll_std_24"] = (
    df.groupby("city")["us_aqi"]
      .transform(lambda x: x.rolling(window=24).std())
)
```

```python
df[
    [
        "city",
        "datetime",
        "us_aqi",
        "aqi_roll_mean_24",
        "aqi_roll_max_24",
        "aqi_roll_min_24",
        "aqi_roll_std_24"
    ]
].head(35)
```

```python
df["target_aqi_24h"] = (
    df.groupby("city")["us_aqi"]
      .shift(-24)
)
```

```python
df[
    [
        "city",
        "datetime",
        "us_aqi",
        "target_aqi_24h"
    ]
].head(30)
```

```python
print("Shape before:", df.shape)

df = df.dropna()

print("Shape after:", df.shape)
```
